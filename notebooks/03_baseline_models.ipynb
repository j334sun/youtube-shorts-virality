# 03 — Exploratory view-count baselines

Linear regression and random forest reproduce the model families explored in the original notebook. The target is `log1p(viewCount)` at observation time. These are not validated predictions of future virality.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
if not (ROOT / "yt_shorts").is_dir():
    raise RuntimeError("Launch Jupyter from the repository root or notebooks/.")
sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from yt_shorts.models import fit_baselines

features = pd.read_csv(ROOT / "data/processed/features.csv")
fitted, metrics = fit_baselines(features)
metrics
```

## Random-forest feature importance
Impurity-based importance can favor certain feature types and is not causal evidence. Interpret it as an exploratory diagnostic.


```python
pipeline = fitted["random_forest"]
forest = pipeline.named_steps["regressor"]
names = pipeline.named_steps["preprocess"].get_feature_names_out()
importance = pd.Series(forest.feature_importances_, index=names).sort_values(ascending=False)
importance.head(10)
```
