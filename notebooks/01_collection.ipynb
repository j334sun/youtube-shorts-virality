# 01 — Collect video metadata

Three collection strategies from the original notebook: category keywords, random common words, and a chosen term. Live requests are off by default. API search results are candidates; they are not verified YouTube Shorts.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
if not (ROOT / "yt_shorts").is_dir():
    raise RuntimeError("Launch Jupyter from the repository root or notebooks/.")
sys.path.insert(0, str(ROOT))
```

```python
from datetime import datetime, timezone
from yt_shorts import collection

RUN_COLLECTION = False
run_label = datetime.now(timezone.utc).strftime("%Y%m%d_%H%M%S")
if RUN_COLLECTION:
    collection.configure_api()
```

## Category keyword queries
The category label describes the query group; the API category ID is a separate field.


```python
if RUN_COLLECTION:
    category_batch = collection.scrape_by_cate(run_label, max_results=10)
```

## Random common-word queries
Supply a word list you may use at `data/resources/words.txt`. The original local word list is not redistributed.


```python
if RUN_COLLECTION:
    random_batch = collection.scrape_by_random(times=50, date=run_label, max_results=5)
```

## A chosen search term
Running all three strategies makes multiple API requests. Choose the cells and request counts deliberately.


```python
if RUN_COLLECTION:
    term_batch = collection.scrape_by_term("foryou", times=10, date=run_label, max_results=25)
```
